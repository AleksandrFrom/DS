# 13. Рисуем линии уровня функциями contour, contourf и tricontour, tricontourf 

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fig = plt.figure(figsize=(7, 4))
ax_3d = fig.add_subplot(projection='3d')

x = np.arange(-2*np.pi, 2*np.pi, 0.2)
y = np.arange(-2*np.pi, 2*np.pi, 0.2)
xgrid, ygrid = np.meshgrid(x, y)

zgrid = np.sin(xgrid) * np.sin(ygrid) / (xgrid * ygrid)

ax_3d.contour(xgrid, ygrid, zgrid) #линии уровей

ax_3d.set_xlabel('x')
ax_3d.set_ylabel('y')
ax_3d.set_zlabel('z')

plt.show()
```

```python
fig, ax = plt.subplots(1, 2)

x = np.arange(-2*np.pi, 2*np.pi, 0.2)
y = np.arange(-2*np.pi, 2*np.pi, 0.2)
xgrid, ygrid = np.meshgrid(x, y)

zgrid = np.sin(xgrid) * np.sin(ygrid) / (xgrid * ygrid)

c1 = ax[0].contour(xgrid, ygrid, zgrid, cmap = 'plasma')
ax[1].contourf(xgrid, ygrid, zgrid)

c1.clabel(colors='k', fmt='%.2f')

plt.show()
```

```python
fig, ax = plt.subplots()

x = np.random.rand(100) * 4*np.pi - 2*np.pi
y = np.random.rand(100) * 4*np.pi - 2*np.pi

z = np.sin(x) * np.sin(y) / (1 + np.abs(x * y))

c1 = ax.tricontour(x, y, z, cmap = 'plasma')

c1.clabel(colors='k', fmt='%.2f')

plt.show()
```
