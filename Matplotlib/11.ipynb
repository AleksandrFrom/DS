# 11. Показ изображений и цветовых сеток

```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
img = Image.open('panda.jpg')
```

```python
fig = plt.figure(figsize=(4, 3))
ax = fig.add_subplot()
ax.imshow(img)

plt.show()
```

```python
plt.imshow(img)
```

```python
data = np.random.randint(0, 255, (100, 100))
plt.imshow(data)

plt.show()
```

```python
plt.imshow(data, cmap='plasma') #цветовая карта cmap

plt.show()
```

```python
img = Image.open('panda.jpg')

b = plt.imshow(img, origin='lower', aspect='equal', alpha=0.7)
plt.colorbar(b)
plt.show()
```

```python
data = np.random.randint(0, 255, (10, 10))

plt.pcolormesh(data) #цветовая сетка

plt.show()
```

```python
data = np.random.randint(0, 255, (10, 10))

b = plt.pcolormesh(data, edgecolors='black', cmap='plasma') #цветовая сетка
plt.colorbar(b)

plt.show()
```
