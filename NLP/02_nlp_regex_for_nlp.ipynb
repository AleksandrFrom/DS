```python
import re
```

```python
chat1 = 'codebasics: you ask lot of questions 😠  1235678912, abc@xyz.com'
chat2 = 'codebasics: here it is: (123)-567-8912, abX_82@xyz.com'
chat3 = 'codebasics: yes, phone: 1235678912 email: abc@xyz.com'
```

```python
pattern1 = '\d{10}'

matches = re.findall(pattern1, chat1)
matches
```

```python
pattern2 = '\(\d{3}\)-\d{3}-\d{4}'

matches = re.findall(pattern2, chat2)
matches
```

```python
pattern3 = '\d{10}|\(\d{3}\)-\d{3}-\d{4}'

matches = re.findall(pattern3, chat1)
matches
```

```python
pattern4 = '[a-z0-9A-Z_]*@[a-z0-9A-Z_]*\.[a-zA-Z]*'

matches = re.findall(pattern4, chat2)
email = matches[0]

email
```

```python
pattern6 = 'age (\d+)'

matches = re.findall(pattern6, text)
order = matches[0]

order
```

```python
pattern7 = 'Born(.*)\n'

matches = re.findall(pattern7, text)
matches[0].strip()
```

```python
pattern8 = 'Born.*\n(.*)\(age'

matches = re.findall(pattern8, text)
matches[0].strip()
```

```python
pattern9 = '\(age.*\n(.*)'

matches = re.findall(pattern9, text)
matches[0].strip()
```

```python
def get_pattern_match(pattern, text):
    matches = re.findall(pattern, text)
    if matches:
        return matches[0]
```

```python
get_pattern_match('\(age.*\n(.*)', text)
```

```python
def get_personal_info(text):
    age = get_pattern_match('age (\d+)', text)
    full_name = get_pattern_match('Born(.*)\n', text)
    birth_date = get_pattern_match('Born.*\n(.*)\(age', text)
    birth_place = get_pattern_match('\(age.*\n(.*)', text)
    return {
        'age': int(age),
        'name': full_name.strip(),
        'birth_date': birth_date.strip(),
        'birth_place': birth_place.strip()
    }
```

```python
get_personal_info(text)
```
