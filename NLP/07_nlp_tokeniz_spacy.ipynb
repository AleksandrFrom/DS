```python
with open("students.txt") as f:
    text = f.readlines()

text
```

```python
text = ' '.join(text)

text
```

```python
#поиск e-mail с помощью spaCy

doc4 = nlp(text)

emails = []

for token in doc4:
    if token.like_email:
        emails.append(token.text)

emails
```

```python
nlp = spacy.blank('ru')

doc = nlp('Шла Саша по шоссе №3 и сосала сушку, в кормане у неё было 200 рублей')

for token in doc:
    print(token, token.is_currency, token.like_num)
```

```python
doc = nlp('gimme double cheese extra large healthy pizza')

tokens = [token.text for token in doc]

tokens
```

```python
from spacy.symbols import ORTH

nlp.tokenizer.add_special_case("gimme", [
    {ORTH: "gim"},
    {ORTH: "me"}
])

doc = nlp('gimme double cheese extra large healthy pizza')

tokens = [token.text for token in doc]

tokens
```

```python
doc = nlp('Dr. Strange loves pav ghaji of mumbai. Hulk loves chat of delhi')

for sentence in doc.sents:
    print(sentence)
```
