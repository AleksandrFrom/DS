```python
def replace_nan(df):
    
    df['Mass size'] = df['Mass size'].fillna(df['Mass size'].mean())
    df['breaking the loop'] = df['breaking the loop'].fillna(df['breaking the loop'].median())
    df['Surface density of the fabric'] = df['Surface density of the fabric'].fillna(df['Surface density of the fabric'].mean())
    df['Prepreg surface density'] = df['Prepreg surface density'].fillna(df['Prepreg surface density'].mean())
    df['Resin content'] = df['Resin content'].fillna(df['Resin content'].mean())
    df['viscosity'] = df['viscosity'].fillna(df['viscosity'].mean())
    df['Gelation time'] = df['Gelation time'].fillna(df['Gelation time'].mean())
    df['Resin Tg'] = df['Resin Tg'].fillna(df['Resin Tg'].mean())
    df['Thickness of the monolayer'] = df['Thickness of the monolayer'].fillna(df['Thickness of the monolayer'].mean())
    df['density'] = df['density'].fillna(df['density'].mean())
    df['Strength_plastik'] = df['Strength_plastik'].fillna(df['Strength_plastik'].median())
    df['Module_plastik '] = df['Module_plastik '].fillna(df['Module_plastik '].median())
    df['LSS'] = df['LSS'].fillna(df['LSS'].mean())
    df['Plastik_Tg'] = df['Plastik_Tg'].fillna(df['Plastik_Tg'].mean())
    
    return df
```

```python
replace_nan(df_144)
df_144
```
