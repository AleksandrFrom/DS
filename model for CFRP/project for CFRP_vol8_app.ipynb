```python
import tkinter as tk
import datetime
import pickle
import numpy as np
```

```python
window = tk.Tk()
window.title('My App')
```

```python
with open('autoclave_model_rf_density.pkl', 'rb') as file:
    density_model = pickle.load(file)

with open('autoclave_model_rf_thikness.pkl', 'rb') as file:
    thikness_model = pickle.load(file)

with open('autoclave_model_rf_strength.pkl', 'rb') as file:
    strength_model = pickle.load(file)

with open('autoclave_model_rf_module.pkl', 'rb') as file:
    module_model = pickle.load(file)

with open('autoclave_model_rf_lss.pkl', 'rb') as file:
    lss_model = pickle.load(file)

with open('autoclave_model_rf_Tg.pkl', 'rb') as file:
    tg_model = pickle.load(file)
```

```python
x = np.array([[190, 1.779, 4.6, 265, 1.7, 1.3, 20, 210, 333, 37.40, 32.33, 12.8, 149.0]])
```

```python
def predict_CFRP(x):
    #x = components()
    x = np.array([x])
    print('Плотность углепластика:', density_model.predict(x))
    print('Толщина монослоя:', thikness_model.predict(x))
```

```python
frame_add_from = tk.Frame(window, bg='green')
frame_statistic = tk.Frame(window, bg='yellow')
frame_list = tk.Frame(window, bg='blue')

frame_add_from.grid(column=0, row=0, sticky='ns')
frame_statistic.grid(column=1, row=0)
frame_list.grid(column=0, row=1, columnspan=2, sticky='we')

l_thiknes_monolayer = tk.Label(frame_statistic, text=thikness_model.predict(x), font='Arial 12 bold')
l_density_pcm = tk.Label(frame_statistic, text='Плотность углепластика, г/см3', font='Arial 12 bold')
l_strenght_pcm = tk.Label(frame_statistic, text='Прочность при растяжении, МПа', font='Arial 12 bold')
l_module_pcm = tk.Label(frame_statistic, text='Модуль упругости при растяжении, ГПа', font='Arial 12 bold')
l_lss_pcm = tk.Label(frame_statistic, text='Прочность при межслоевом сдвиге, МПа', font='Arial 12 bold')
l_tg_pcm = tk.Label(frame_statistic, text='Температура стеклования, оС', font='Arial 12 bold')

l_thiknes_monolayer.grid(row='0', column='0', sticky='w', padx=10, pady=10)
l_density_pcm.grid(row='0', column='1', sticky='e', padx=10, pady=10)
l_strenght_pcm.grid(row='1', column='0', sticky='w', padx=10, pady=10)
l_module_pcm.grid(row='1', column='1', sticky='e', padx=10, pady=10)
l_lss_pcm.grid(row='2', column='0', sticky='w', padx=10, pady=10)
l_tg_pcm.grid(row='2', column='1', sticky='e', padx=10, pady=10)

l_temp_frame_add_form = tk.Label(frame_add_from, text='frame_add_form')
l_temp_frame_list = tk.Label(frame_list, text='frame_list')
l_temp_frame_add_form.pack(expand=True, padx=20, pady=20)
l_temp_frame_list.pack(expand=True, padx=20, pady=20)

window.mainloop()
```
