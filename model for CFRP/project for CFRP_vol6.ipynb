```python
import pandas as pd
import numpy as np

# инструменты для построения модели:
from sklearn.linear_model import LinearRegression # инструмент для создания и обучения модели
from sklearn.ensemble import RandomForestRegressor # инструмент для создания и обучения модели
from xgboost import XGBRegressor

import pickle
```

```python
with open('autoclave_model_rf_density.pkl', 'rb') as file:
    density_model = pickle.load(file)

with open('autoclave_model_rf_thikness.pkl', 'rb') as file:
    thikness_model = pickle.load(file)
```

```python
def components(**kwargs):
    X = []
    tex = input('линейная плотность УВ') 
    X.append(float(tex))
    density_yarn = input('объемная плотность УВ')
    X.append(float(density_yarn))
    strenght = input('прочность УВ')
    X.append(float(strenght))
    module = input('модуль упругости УВ') 
    X.append(float(module)) 
    lengthening = input('удлинение при разрыве УВ')
    X.append(float(lengthening))  
    size = input('содержание аппрета на УВ')
    X.append(float(size))  
    loop = input('разрыв в петле') 
    X.append(float(loop)) 
    fabric_surf_dens = input('поверхностная плотность ткани')
    X.append(float(fabric_surf_dens)) 
    prepreg_surf_dens = input('поверхностная плотность препрега')
    X.append(float(prepreg_surf_dens)) 
    resin_cont = input('содержание связующего в препреге')
    X.append(float(resin_cont))
    viscosity = input('вязкость связующего')
    X.append(float(viscosity))
    gel_time = input('время гелеобразования связующего')
    X.append(float(gel_time))
    res_Tg = input('температура стеклования связующего')
    X.append(float(res_Tg))
    return X
```

```python
def predict_CFRP(x):
    x = components()
    x = np.array([x])
    print('Плотность углепластика:', density_model.predict(x))
    print('Толщина монослоя:', thikness_model.predict(x))
    
```

```python
predict_CFRP(x)
```
