```python
import tkinter as tk
```

```python
#метод grid

window = tk.Tk()
window.title('My app')

frame_add_form = tk.Frame(window, width = 150, height = 150, bg = 'green')
frame_statistik = tk.Frame(width = 150, height = 150, bg = 'yellow')
frame_list = tk.Frame(width = 300, height = 150, bg = 'blue')

frame_add_form.grid(row=0, column=0)
frame_statistik.grid(row=0, column=1)
frame_list.grid(row=1, column=0, columnspan=2)

window.mainloop()
```

```python
#метод place

window = tk.Tk()
window.title('My app')
window.geometry('300x300')
#window.minsize(290, 290)
#window.maxsize(310, 310)
#window.resizable(width=False, height=False)

frame_add_form = tk.Frame(window, width = 150, height = 150, bg = 'green')
frame_statistik = tk.Frame(width = 150, height = 150, bg = 'yellow')
frame_list = tk.Frame(width = 300, height = 150, bg = 'blue')

frame_add_form.place(relx = 0, rely = 0, relheight = 0.5, relwidth = 0.5)
frame_statistik.place(relx = 0.5, rely = 0, relheight = 0.5, relwidth = 0.5)
frame_list.place(relx = 0, rely = 0.5, relheight = 0.5, relwidth = 1)

window.mainloop()
```

```python
#метод pack

window = tk.Tk()
window.title('My app')
window.geometry('300x300')
#window.minsize(290, 290)
#window.maxsize(310, 310)
#window.resizable(width=False, height=False)

frame_wrap = tk.Frame(window)
frame_add_form = tk.Frame(frame_wrap, bg = 'green')
frame_statistik = tk.Frame(frame_wrap, bg = 'yellow')
frame_list = tk.Frame(frame_wrap, bg = 'blue')

frame_wrap.pack(fill='both')
frame_add_form.pack(side='left', fill='both', expand=True, ipady=60)
frame_statistik.pack(side='right', fill='both', expand=True, ipady=60)
frame_list.pack(side='bottom', fill='both', expand=True)

window.mainloop()
```
