```python
# Replace 'Not provide' in 'Employee Name' column to NaN

df_droped['EmployeeName'] = df_droped['EmployeeName'].replace('Not provided', np.nan)
```

```python
#Drop Raws having 5 missing values

df_droped.drop(df_droped[df_droped.isnull().sum(axis=1) == 5].index, axis=0, inplace=True)
```
